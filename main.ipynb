```python
import os
import requests
from dotenv import load_dotenv

# Carrega as variáveis do arquivo .env
load_dotenv()

TOKEN = os.getenv("token")

# Endereço base da API

URL_BASE = "https://api.olhovivo.sptrans.com.br/v2.1"

# Cria uma sessão para manter a autenticação

sessao = requests.Session()

# Verifica se o token existe

def autenticar():
    if not TOKEN:
        print("Token não encontrado!")
    else:
        # Realiza a autenticação
        resposta = sessao.post(
            f"{URL_BASE}/Login/Autenticar",
        params={"token": TOKEN}
    )

    if resposta.status_code == 200 and resposta.json() is True:
        print("Autenticação realizada com sucesso!")

        # Primeira consulta: buscar uma linha
        consulta = sessao.get(
            f"{URL_BASE}/Linha/Buscar",
            params={"termosBusca": "8000"}
        )

        if consulta.status_code == 200:
            linhas = consulta.json()

            print("\nLinhas encontradas:")

            for linha in linhas:
                print(linha["lt"], linha["tp"], "-", linha["ts"])
        else:
            print("Erro ao consultar linhas:", consulta.status_code)

    else:
        print("Falha na autenticação!")
```
